# Traffic Accident Data — Exploratory Data Analysis

This notebook walks through the full EDA pipeline:
1. Data Loading (PySpark)
2. Data Cleaning
3. Core Analysis
4. Visualization

```python
import sys
sys.path.insert(0, '../src')

from data_loader import create_spark_session, load_csv, show_sample
from data_cleaning import clean_data
from analysis import run_all_analysis
from visualization import generate_all_static_charts

import warnings
warnings.filterwarnings('ignore')
```

## 1. Load Data

```python
spark = create_spark_session('EDA')
df_raw = load_csv(spark, '../data/accidents.csv')
show_sample(df_raw)
```

## 2. Clean Data

```python
df_clean = clean_data(df_raw)
df_clean.show(5)
print('Columns:', df_clean.columns)
```

## 3. Run Analysis

```python
results = run_all_analysis(df_clean)

print('\n--- Accidents by City ---')
print(results['by_city'].to_string(index=False))

print('\n--- Severity Distribution ---')
print(results['severity'].to_string(index=False))

print('\n--- Peak Accident Hours ---')
print(results['peak_hours'].to_string(index=False))
```

## 4. Spark SQL Examples

```python
sql = results['sql_results']
print('High severity by city:')
print(sql['high_severity_by_city'].to_string(index=False))
```

## 5. Generate Static Charts

```python
chart_paths = generate_all_static_charts(results)
for name, path in chart_paths.items():
    print(f'{name}: {path}')
```

```python
from IPython.display import Image, display
for path in chart_paths.values():
    display(Image(filename=path, width=800))
```
